## Personal Story

In 2025, I faced a real decision: relocating my family from Northern Ireland to London for better career opportunities.

As a father, safety and school quality mattered most. As a PSV Mechanic transitioning into Data Analytics, I decided to treat this as a data project — not rely on guesswork or estate agent recommendations.

**London has 32 boroughs. Each varies widely in crime, school quality, housing cost, and liveability.**

This project documents how I used publicly available data to find the safest, most family-friendly boroughs.

---

## Project Objectives

- **Primary question:** Which London boroughs are safest for families with young children?
- **Key criteria:** Low violent crime, quality schools (Ofsted), affordable housing, reasonable commute

## Dataset

- **Source:** [data.police.uk](https://data.police.uk) public API (no key required)
- **Size:** 611,000+ crime incidents
- **Coverage:** 33 London boroughs, 24 months (2024-2025)
- **Crime types:** 15 categories including violence, burglary, drugs, ASB
